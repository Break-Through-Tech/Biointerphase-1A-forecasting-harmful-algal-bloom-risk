```python
# Combine the year, month, and day columns into one datetime column
bradentonFL["date"] = pd.to_datetime(bradentonFL[["year", "month", "day"]])
```

```python
# Calculate the average daily temperature
bradentonFL["avg_temp"] = (bradentonFL["max_temp"] + bradentonFL["min_temp"]) / 2
```

```python
# Calculate the difference between the maximum and minimum temperatures
bradentonFL["temp_range"] = bradentonFL["max_temp"] - bradentonFL["min_temp"]
```

```python
# Create an interaction between maximum and minimum temperature
bradentonFL["temp_interaction"] = bradentonFL["max_temp"] * bradentonFL["min_temp"]
```

```python
# Create a binary feature indicating whether any precipitation occurred
bradentonFL["rained"] = (bradentonFL["precipitation"] > 0).astype(int)
```

```python
# Convert the weather station ID into a categorical variable
bradentonFL["coop_id"] = bradentonFL["coop_id"].astype(str)
```

```python
# Divide average temperature into bins
bradentonFL["temp_bin"] = pd.cut(
    bradentonFL["avg_temp"],
    bins=[-np.inf, 50, 70, 85, np.inf],
    labels=["cold", "mild", "warm", "hot"]
)
```

```python
# coop_id: unnecessary if this is only one weather station
# day: redundant because date already contains the day
bradentonFL = bradentonFL.drop(columns=["coop_id", "day"], errors="ignore")
```

```python
# Standardizing continuous weather measurements.

sarasota_numeric = [
    "air_pressure_at_mean_sea_level",
    "dew_point_temperature",
    "air_temperature",
    "visibility_in_air",
    "wind_speed_of_gust",
    "wind_speed"
]
sarasota_numeric = [c for c in sarasota_numeric if c in sarasota.columns]

sarasota_scaler = StandardScaler()
sarasota[sarasota_numeric] = sarasota_scaler.fit_transform(sarasota[sarasota_numeric])
```

Circular-encoding wind direction. 359 degrees and 1 degree are not very "close" numbers, but roughly the same degree. We split this part of the data into sine/cosine.

```python
if "wind_from_direction" in sarasota.columns:
    wind_rad = np.radians(sarasota["wind_from_direction"])
    sarasota["wind_dir_sin"] = np.sin(wind_rad)
    sarasota["wind_dir_cos"] = np.cos(wind_rad)
    sarasota = sarasota.drop(columns=["wind_from_direction"])
```

A raw timestamp string won't be useful for our model because it hides patterns relating to season that are important for what we are trying to accomplish. Month is cyclically encoded (similar to what was done with wind direction).

```python
if "time" in sarasota.columns:
    dt = pd.to_datetime(sarasota["time"], utc=True, errors="coerce")
    sarasota["month"] = dt.dt.month
    sarasota["month_sin"] = np.sin(2 * np.pi * sarasota["month"] / 12)
    sarasota["month_cos"] = np.cos(2 * np.pi * sarasota["month"] / 12)
```

Standardizing sea_water-temperature.

```python
mote_numeric = ["sea_water_temperature"]

motemarine[mote_numeric] = motemarine[mote_numeric].apply(pd.to_numeric, errors='coerce')
motemarine.dropna(subset=mote_numeric, inplace=True)

mote_scaler = StandardScaler()
motemarine[mote_numeric] = mote_scaler.fit_transform(motemarine[mote_numeric])
```

Similar to the Sarasota dataset, here we cyclically encode month from the timestamp. Seasonality is important for determining bloom risk. This keeps January and December "close" together.

```python
if "time" in motemarine.columns:
    dt = pd.to_datetime(motemarine["time"], utc=True, errors="coerce")
    motemarine["month"] = dt.dt.month
    motemarine["month_sin"] = np.sin(2 * np.pi * motemarine["month"] / 12)
    motemarine["month_cos"] = np.cos(2 * np.pi * motemarine["month"] / 12)
```
